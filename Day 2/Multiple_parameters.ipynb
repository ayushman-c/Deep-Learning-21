### 1. Dataset Preparation
We define our input feature matrix $X$ (with 2 features per sample) and the target values $y$ for training our multiple linear regression model.

```python
import numpy as np

X = np.array([
    [1, 5],
    [2, 6],
    [3, 5],
    [4, 7],
    [5, 6],
    [6, 8],
    [7, 7],
], dtype=float)

y = np.array([45, 54, 58, 68, 72, 82, 86], dtype=float)
```

### 2. Parameter Initialization
We initialize our weights vector $w$ and bias $b$ to 0.0, set a learning rate of $0.01$, and specify $1,000,000$ epochs for gradient descent.

```python
w = np.array([0.0, 0.0])
b = 0.0

learning_rate = 0.01
epoch = 1000000
```

### 3. Model Training via Gradient Descent
We optimize the weights $w$ and bias $b$ by iteratively calculating the Mean Squared Error (MSE) loss, computing gradients, and updating parameters.

```python
for i in range(epoch):
  y_pred = X @ w +b
  loss = np.mean((y - y_pred)**2)

  dw = (-2/len(X)) * (X.T @ (y - y_pred))
  db = (-2/len(X)) * np.sum(y - y_pred)

  w = w - learning_rate * dw
  b = b - learning_rate * db


print(f"Epoch: {i}, Loss: {loss}, w : {w}, b : {b}")
```
