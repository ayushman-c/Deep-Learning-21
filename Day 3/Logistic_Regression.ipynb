### Part 1: Logistic Regression Training

This section initializes the model parameters and trains a binary classifier using **Gradient Descent**.

#### Key Components:
* **Sigmoid Activation Function**: Maps any real-valued number into a probability range $(0, 1)$:
  $$\sigma(z) = \frac{1}{1 + e^{-z}}$$
* **Loss Function (Binary Cross-Entropy)**: Measures the error between predicted probabilities $p$ and actual targets $y$:
  $$J(w, b) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(p^{(i)}) + (1 - y^{(i)}) \log(1 - p^{(i)}) \right]$$
* **Gradient Calculation**: Computes partial derivatives of the loss with respect to the weight $w$ and bias $b$:
  $$dw = \frac{1}{m} X^T (p - y)$$
  $$db = \frac{1}{m} \sum (p - y)$$
* **Gradient Descent Update**: Iteratively updates parameters using the learning rate ($lr$):
  $$w = w - lr \cdot dw$$
  $$b = b - lr \cdot db$$

```python
import numpy as np

def sigmoid(z):
  return 1 / (1 + np.exp(-z))

X = np.array([1, 2, 3, 4, 5])
y = np.array([0, 0, 0, 1, 1])

w = 1.0
b = -3.0
lr =0.01


for epoch in range(1000):
  z = w *X +b
  p = sigmoid(z)

  loss = -(1/len(y))*np.sum(y*np.log(p) + (1-y)*np.log(1-p))

  dw = (1/len(y)) * (X.T @ (p-y))
  db = (1/len(y)) * np.sum(p-y)

  w = w - lr*dw
  b = b - lr*db

  if epoch % 100 == 0:
    print(f"Epoch {epoch}, Loss: {loss}, w: {w}, b: {b}")

```

### Part 2: Model Evaluation and Inference

Once the training loop finishes and finds the optimal weight ($w$) and bias ($b$), we evaluate how well the model has fit the training data.

#### Evaluation Workflow:
1. **Compute Log-odds ($z$)**: Multiply the features $X$ by the learned weight $w$ and add the bias $b$.
2. **Generate Probabilities ($p$)**: Pass $z$ through the sigmoid function to get values between 0 and 1.
3. **Decision Threshold**: Apply a threshold (usually $0.5$) to classify outputs as either class `0` or class `1`.
4. **Compare Outputs**: Display the final probabilities and check the matching rate between the binary predictions and the actual labels.

```python
z = w * X + b
p = sigmoid(z)

predictions = (p >= 0.5).astype(int)

print("Final w:", w)
print("Final b:", b)
print("Probabilities:", p)
print("Predictions:", predictions)
print("Actual:", y)
```
