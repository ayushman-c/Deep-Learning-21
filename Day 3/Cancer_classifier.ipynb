# Logistic Regression from Scratch on Breast Cancer Dataset

This notebook demonstrates how to implement a binary **Logistic Regression** classifier from scratch using NumPy. We will load the dataset, normalize our features, define the necessary mathematical functions, run gradient descent to train the model, and evaluate its performance.

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

data = load_breast_cancer()

X = data.data
y = data.target

lr = 0.001

print(X.shape)
print(y.shape)

np.unique(y, return_counts=True)
```

## 1. Dataset Partitioning

We split our data into training (80%) and testing (20%) sets to ensure we can evaluate our model on unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=69)

```

## 2. Activation Function

The Logistic Regression model uses the **Sigmoid function** to map any real-valued number into a probability between 0 and 1:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

## 3. Feature Scaling & Initialization

To help gradient descent converge faster and prevent numerical instability, we standardize our features to have a mean of 0 and a standard deviation of 1. We also initialize our weights ($w$) to zeros and bias ($b$) to 0.0.

```python
mean_X = np.mean(X_train, axis=0)
std_X = np.std(X_train, axis=0)

X_train_scaled = (X_train - mean_X)/std_X
X_test = (X_test - mean_X)/std_X

w = np.zeros(X_train_scaled.shape[1])
b = 0.0
```

## 4. Model Training with Gradient Descent

We optimize the weights $w$ and bias $b$ using the binary cross-entropy loss function. In each epoch, we:
1. Compute the linear combination: $z = XW + b$
2. Predict probabilities: $p = \sigma(z)$
3. Calculate Binary Cross-Entropy Loss: $L = -\frac{1}{m}\sum[y\log(p) + (1-y)\log(1-p)]$
4. Compute gradients: $dw = \frac{1}{m}X^T(p - y)$ and $db = \frac{1}{m}\sum(p-y)$
5. Update parameters: $w = w - \alpha \cdot dw$ and $b = b - \alpha \cdot db$ where $\alpha$ is our learning rate.

```python

w = np.zeros(X_train_scaled.shape[1])
b = 0.0


for epoch in range(10000):

  z = X_train_scaled @ w + b
  p = sigmoid(z)


  p = np.clip(p, 1e-15, 1 - 1e-15)

  loss = -(1 / len(y_train)) * np.sum(
      y_train * np.log(p) +
      (1 - y_train) * np.log(1 - p)
  )

  dz = p - y_train
  dw = (1 / len(y_train)) * X_train_scaled.T @ dz
  db = (1 / len(y_train)) * np.sum(dz)

  w = w - lr * dw
  b = b - lr * db

  if epoch % 1000 == 0:
      print(f"Epoch {epoch}, Loss: {loss:.4f}")

print(f"Final Loss: {loss:.4f}")
```

## 5. Evaluation

Now we can evaluate how well our trained weights perform on the unseen test set.

```python

z_test = X_test @ w + b
p_test = sigmoid(z_test)
y_pred = (p_test >= 0.5).astype(int)


accuracy = np.mean(y_pred == y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```
