# Simple Linear Regression from Scratch
This notebook demonstrates the implementation of a basic Linear Regression model using Gradient Descent to find the relationship between a feature set `x` and target labels `y`.

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 8, 11, 14, 17], dtype=float)
```

## Data Preparation
We define a small synthetic dataset where the relationship follows the linear equation $y = 3x + 2$. The goal of our model is to learn these weights (w=3, b=2) from the data.

# Defining parameters

```python
w = 0.0
b = 0.0

learning_rate = 0.01

# our equation y_hat = w*x + b
```

## Model Training (Gradient Descent)
We optimize the Mean Squared Error (MSE) loss function using the following gradients:
- $\frac{\partial L}{\partial w} = \frac{2}{n} \sum x(y_{pred} - y)$
- $\frac{\partial L}{\partial b} = \frac{2}{n} \sum (y_{pred} - y)$

We iteratively update $w$ and $b$ to minimize the loss over 3000 epochs.

```python
for i in range(3000):

  y_pred = w*x + b

  loss = np.mean((y - y_pred)**2)

  dw = (2 / len(x)) * np.sum(x * (y_pred - y))
  db = (2 / len(x)) * np.sum(y_pred - y)

  w = w - learning_rate * dw
  b = b - learning_rate * db


  if i % 100 == 0:
        print(f"Epoch: {i}, Loss: {loss:.4f}, w: {w:.4f}, b: {b:.4f}")
```
