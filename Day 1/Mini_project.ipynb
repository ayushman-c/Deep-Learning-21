```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5], dtype=float)
y = np.array([12,19,26,33,40], dtype=float)

#equation y = 7x + 5
```

```python
w = 0.0
b = 0.0

lr = 0.01
```

```python
for i in range (2000):

  y_pred = w*x + b

  loss = np.mean((y - y_pred) ** 2)

  dw = (2/len(x)) * np.sum((y_pred - y)*x)
  db = (2/len(x)) * np.sum((y_pred - y))

  w = w - lr*dw
  b = b - lr*db

  if i%100 == 0:
   print(f"Interation: {i}, Loss: {loss:.4f}, w: {w:.4f}, b: {b:.4f}")
```
